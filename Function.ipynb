```python
# Write a python function that takes a list of numbers as input and returns the sum of all even numbers in the list.
```

```python
def sum_of_even_numbers(numbers):
    return sum(num for num in numbers if num % 2 == 0)

my_list = [1, 2, 3, 4, 5, 6]
result = sum_of_even_numbers(my_list)
print("Sum of even numbers:", result)
```

```python
# Write a python function that checks if a given number is prime or not from 1 to 200.
```

```python
def is_prime(n):
    if n < 2 or n > 200:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

for num in range(1, 201):
    if is_prime(num):
        print(f"{num} is a prime number")
```

```python
# Create an iterator class in python that generates the Fibonacci sequence up to a specified number of terms.
```

```python
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.a


fib = FibonacciIterator(10)
for num in fib:
    print(num)
```

```python
# Write a generator function in Python that yields the powers of 2 up to a given exponent.
```

```python
def powers_of_two(max_exponent):
    for exp in range(max_exponent + 1):
        yield 2 ** exp

for value in powers_of_two(5):
    print(value)
```

```python
# Implement a generator function that reads a file line by line and yields each line as a string.
```

```python
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.rstrip('\n')
```

```python
# Write a python program that uses 'map()' to to convert a list of temperatures from Celsius to Fahrenheit.
```

```python
def c_to_f(celsius):
    return (celsius * 9/5) + 32

celsius_temps = [0, 20, 30, 37, 100]

fahrenheit_temps = list(map(c_to_f, celsius_temps))

print("Temperatures in Fahrenheit:", fahrenheit_temps)
```

```python
# Create a python program that uses 'filter()' to remove all the vowels from a given string.
```

```python
def remove_vowels(input_str):
    vowels = 'aeiouAEIOU'
    return ''.join(filter(lambda char: char not in vowels, input_str))

text = "Hello, World!"
result = remove_vowels(text)
print("String without vowels:", result)
```

Q.1. What is the diffrence between a function and a method in Python?
--> A function is a block of code that performs a specific task and can be called independently. It is not bound to any object or class.

A method is a function that is associated with an object or class. It operates on the data (attributes) contained within that object and is called on an instance of a class.

# Function Example
def greet(name):
    return f"Hello, {name}!"

# Method Example (Part of a class)
class Person:
    def __init__(self, name):
        self.name = name
    
    def greet(self):
        return f"Hello, {self.name}!"

# Using the function
print(greet("Alice"))  # Output: Hello, Alice!

# Using the method (called on an object)
person = Person("Bob")
print(person.greet())  # Output: Hello, Bob!


Q.2. Explain the concept of function arguments and paramaeters in Python.

--> Parameters are the variables defined inside the parentheses of a function definition. They act as placeholders for the values that will be passed to the function when it is called.

def greet(name):  # 'name' is the parameter
    print(f"Hello, {name}!")

Arguments are the actual values that are passed to the function when it is called. These values correspond to the parameters defined in the function.

greet("Alice")  # "Alice" is the argument


Q.3. What are the diffrent ways to define and call a function in python?

--> 1. Basic Function Definition and Call
This is the most common way to define a function using the def keyword and calling it by its name.

# Function Definition
def greet(name):
    print(f"Hello, {name}!")

# Function Call
greet("Alice")


2. Function with Default Arguments
You can define a function with default argument values. If the caller doesn't provide a value for the argument, the default value will be used.

def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()        # Uses default argument
greet("Bob")   # Overrides default argument


3. Function with Variable-Length Arguments (*args and **kwargs)
You can define functions to accept a variable number of positional (*args) or keyword arguments (**kwargs).

def print_numbers(*args):
    for number in args:
        print(number)

print_numbers(1, 2, 3, 4)  # Positional arguments passed as a tuple

def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=25)  # Keyword arguments passed as a dictionary


4. Lambda Functions (Anonymous Functions)
You can define functions in a single line using lambda. These are anonymous functions, often used when you need a simple function for a short period.

# Lambda function that adds two numbers
add = lambda a, b: a + b
print(add(3, 4))  # Output: 7

5. Function as an Argument (Higher-Order Functions)
You can pass functions as arguments to other functions.

def apply_operation(func, a, b):
    return func(a, b)

def multiply(x, y):
    return x * y

result = apply_operation(multiply, 3, 4)
print(result)  # Output: 12


6. Function Returning Another Function
You can define a function that returns another function. This is useful for creating function factories or closures.

def outer_function(message):
    def inner_function():
        print(message)
    return inner_function

greet = outer_function("Hello, World!")
greet()  # Output: Hello, World!

Q.5. What are iterators in python and how do the differs from iterables?

--> In Python, an iterable is any object that can be iterated over, like lists or strings, and implements the __iter__() method. An iterator is an object that allows iteration over an iterable and implements both __iter__() and __next__() methods. The iterator keeps track of the iteration state and produces the next item in the sequence, raising StopIteration when no more items are left. An iterable is used to create an iterator.
